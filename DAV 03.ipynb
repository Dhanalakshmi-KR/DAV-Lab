```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sms
```

```python
df=pd.read_csv("student_data.csv")
```

```python
plt.hist(df['age'], bins=10)
plt.xlabel('Age')
plt.ylabel('Number of Students')
plt.title('Age Distribution of Students')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(x=df['age'])
plt.title('Boxplot of Student Age')
plt.xlabel('Age')
plt.show()
```

```python
sns.histplot(df['studytime'], bins=4, kde=False)
plt.title('Distribution of Study Time')
plt.xlabel('Study Time')
plt.ylabel('Number of Students')
plt.show()
```

```python
sns.histplot(df['traveltime'], bins=4, kde=False)
plt.title('Distribution of Travel Time')
plt.xlabel('Travel Time')
plt.ylabel('Number of Students')
plt.show()
```

```python
sns.histplot(df['failures'], bins=4, kde=False)
plt.title('Distribution of Failures')
plt.xlabel('Number of Failures')
plt.ylabel('Number of Students')
plt.show()
```

```python
sns.histplot(df['famsize'], kde=False)
plt.title('Distribution of Family Size')
plt.xlabel('Family Size')
plt.ylabel('Number of Students')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

sns.histplot(df['age'], bins=10, kde=True, ax=axes[0, 0])
axes[0, 0].set_title('Age Distribution')

sns.histplot(df['traveltime'], bins=4, kde=False, ax=axes[0, 1])
axes[0, 1].set_title('Travel Time Distribution')

sns.histplot(df['studytime'], bins=4, kde=False, ax=axes[1, 0])
axes[1, 0].set_title('Study Time Distribution')

sns.histplot(df['failures'], bins=4, kde=False, ax=axes[1, 1])
axes[1, 1].set_title('Failures Distribution')

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(x=df['age'], y=df['studytime'])
plt.title('Age vs Study Time')
plt.xlabel('Age')
plt.ylabel('Study Time')
plt.show()
```
